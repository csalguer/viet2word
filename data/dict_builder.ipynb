```python
import pandas as pd
import os
import unicodedata
import re

from tqdm import tqdm
```

```python
path = os.getcwd()
```

```python
files = ['Chinese Hanviet Cognates.tsv', 'Chinese Hanviet Non-Cognates.tsv', 'Hanviet Same Sound.tsv', 'Nom Standardization.csv', 'Vietnamese English Dictionary.csv']
```

```python

ret = {}
with open(path+f'/vn-dict/{files[-1]}') as f:
  lines = f.readlines()
  curr = {}
  for line in lines:
    if(line[0] == '@'):
      curr["headword"] = line[1:] 
    if(line[0] == '*'):
      curr['partOfSpeech'] = line[1:].strip()
    if(line[0] == '-'):
      defn = line[1:]
      if "+" in defn:
        definition, example = defn.split("+").trim()
      if curr["definitions"]:
        curr["definitions"].extend(list({"definition": definition, "example": example}))
      else:
        curr["definitions"] = list({"definition": definition, "example": example}) 

  ret.append(curr)
  curr.clear
```
